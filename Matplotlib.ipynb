```python
oscar_movies={
    "hero no 1",
    "biwi no 1",
    "jai baba ki",
    "krrish",
    "jai ho"
}
years=[2008,2009,2010,2011,2012]
oscar_revenues=[1005,170,427,133,232] #in M$

#plot
plt.plot(years,oscar_revenues)
plt.title("Oscar movies revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue in million dollar")
```

```python
non_oscar_movies={
    "Judwa2","coolie n0. 1" ,"Boss","Sardar","Ganpat"
}

non_oscar_revenue=[378,2788,829,185,275]

plt.plot(years,oscar_revenues,label='Oscar Movies')
plt.plot(years,non_oscar_revenue, label='Non Oscar Movies')
plt.title(" Oscar Movies vs Non Oscar Movies")
plt.xlabel("Year")
plt.ylabel("Revenue")

plt.legend(loc='upper right')  # it wil tell labels of these two different graphs
```
