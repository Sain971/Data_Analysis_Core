```python
oscar_movies={
    "hero no 1",
    "biwi no 1",
    "jai baba ki",
    "krrish",
    "jai ho"
}
years=[2008,2009,2010,2011,2012]
oscar_revenues=[1005,170,427,133,232]  

plt.bar(years,oscar_revenues,color='purple')
plt.title("Revene in each year")
plt.xlabel("Movies")
plt.ylabel("Revenue")
for i in range(len(years)):
    plt.text(years[i],oscar_revenues[i]+20,str(oscar_revenues[i]),ha='center')
plt.ylim(0,max(oscar_revenues)+100)# to increase the border difference in x and y axis
```

Multiple datasets on Bar Chart

```python
non_oscar_movies={
    "Judwa2","coolie n0. 1" ,"Boss","Sardar","Ganpat"
}

non_oscar_revenue=[378,2788,829,185,275]

#compare 2 sets of data 

plt.bar(years,oscar_revenues)
plt.bar(years,non_oscar_revenue)

plt.title("Oscar vs Non-oscar Movies Revenue")

plt.xlabel("Year")
plt.ylabel("Revenue")
```

```python
import numpy as np
oscar_movies={
    "hero no 1",
    "biwi no 1",
    "jai baba ki",
    "krrish",
    "jai ho"
}
years=[2008,2009,2010,2011,2012]
oscar_revenues=[1005,170,427,133,232]  

non_oscar_movies={
    "Judwa2","coolie n0. 1" ,"Boss","Sardar","Ganpat"
}

non_oscar_revenue=[378,2788,829,185,275]

#compare 2 sets of data 
x= np.arange(len(years))
# x = [0, 1, 2, 3, 4] Ye actual years nahi hain. Ye sirf graph ke x-axis par positions hain.
width=0.4

plt.bar(x-width/2,oscar_revenues,width,label="oscar movies")
plt.bar(x+width/2,non_oscar_revenue,width,label="non oscar mocies")

 
plt.title("Oscar vs Non-oscar Movies Revenue")

plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()

```
