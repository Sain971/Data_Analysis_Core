```python
import numpy as np
import pandas as pd
"""
*find-->    df.isna(),df.isna().sum(),df.isna().any()
*remove---> df.dropna(thresh=3)
*filling--> df.fillna()

"""
#nothing is applying permanently to do so use (inplace=True)
```

Finding Missing data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
#nan is null values
df=pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()#it will sum of all null values in our data in a colomn
```

```python
df.isna().any()#check will there is null value or not 
```

REMOVING NULL VALUES

```python
df
```

```python
df.dropna()#poora row delete ho jayega jisme bhi null value hoga
```

```python
df.dropna(thresh=3)#agar 3 nulll values hongi toh row will be removed
```

FILLING THE MISSING DATA

```python
df
```

```python
df.fillna(0)
```

```python
#diff values for each coloumn
v={"A":0,"B":100,"C":233,"D":144}
df.fillna(value=v)
```

```python
df.fillna(df.mean())
```
