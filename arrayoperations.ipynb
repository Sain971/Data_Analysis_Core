Deep And Shallow Copy

```python
a=np.arange(1,20)
a
```

```python
slice=a[0:5]
slice
```

```python
slice=slice*10
slice
#Yaha pei toh change hua but ye change apna main array mei koi change mhi krega 
#magar agar apnei paas ek list hoti toh original list mei bhii change ho jata 
"""THIS IS THE CONCEPT OF SHALLOW COPY"""
#a arary and slice array are at diff places in ram
```

```python
b=a
#change in b will effect change in a cause they both are in same place at ram 
b[0]=99
a
"""This is Deep copy 
to avoid this we can use b=a.copy() it will not affect a elements
"""

```

Advance ArraY Manipulation

```python
"""Stacking Array"""
#We can't stack two matrix 
# We can only stack two vectors or Two Arrays
nums1=np.array([1,2,3,4,5])
nums2=np.array([5,6,7,8,9])
```

```python
#Vertical stack
np.vstack((nums1,nums2))
```

```python
#Vertical Stack
np.hstack((nums1,nums2))
```

```python
#coloumn satck
np.column_stack((nums1,nums2))
```
