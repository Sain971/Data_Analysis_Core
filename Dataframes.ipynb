```python
import numpy as np
import pandas as pd
```

## Creating a Data-frame

```python
#creating DF from a DICT
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
#creating Df from a LIST
data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2=pd.DataFrame(data_list)
col_name=["Name","Age","City","Salary"]
df2=pd.DataFrame(data_list,columns=col_name)
```

```python
df2
```

```python
#axis =0 #vertical axis 
#axis =1 #horizontal axis
"""
      Columns →
      A   B   C
Rows ↓
0
1
2
"""
#df.drop will not delete our coloumn permanently it will just show the temporary look 
#for permanent delete we have to use df.drop(["Salary",axis=1,inplace=true])
df2.drop("Salary",axis=1,inplace=True)
# for multiple delete just use list  df.drop(["name","age"],axis=1,inplace=True)
```
