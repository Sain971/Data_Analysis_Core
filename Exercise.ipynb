Valid Suduko

```python
import numpy as np
```

```python
#In suduko there are 9 3by3 blocks
#Every rows and coloumns should give a sum of 45i.e 1+.....+9
#Every block should also have a sum of 45
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

```python
#sum of each and every row
b=np.sum(s,axis=1)
#checking
b==45
#or 
for i in b:
    if i!=45:
        break 
else:
    print("for rows it is valid")
```

```python
#sum of each coloumn
c=np.sum(s,axis=0)
for i in c:
    if i!=45:
        break
else:
    print("For coloumns it is valid")

```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students.
age=data[::,:1:]
age.mean()
```

```python
#Extract Math marks of all students.
math_marks=data[::,1:2]
print(f"The marks of math is :\n{math_marks}")
```

```python
#Find the highest Science mark.
max_science=data[::,2::]
print(f"highest Science mark: {max_science.max()}")
```

```python
#Get details of the student who scored more than 90 in Math.
details=data[data[::,1]>90]
details
```

```python
#Increase Math marks of all students by 5.
data[:,1]+=5
data
```

```python
#Find how many students are younger than 19.
younger=data[data[:,0]<19]
len(younger)
```

```python
#Calculate the average marks in each subject (column-wise mean).
coloumnmean=data[:,1:]
np.mean(coloumnmean,axis=0)
```

```python
#Get data of students who scored at least 80 in both subjects.
data[(data[:,1]>=80) & (data[:,2]>=80)]
#use this when there is a condition that we have to apply to filter our data
```
